```python
'''
Question 1 : What is React?
Answer:  React is a free and open-source JavaScript library developed by Meta (Facebook) for building user interfaces, particularly single-page applications (SPAs). It relies on a component-based architecture, allowing developers to create reusable UI elements. React maintains a Virtual DOM in memory to calculate minimal updates efficiently before updating the real DOM.


Question 2 : Write a React functional component named `Greeting` that accepts a `name` prop and displays: "Hello, [name]!”.
Answer: function Greeting({ name }) {
  return <h1>Hello, {name}!</h1>;
}

export default Greeting;


Question 3 : What is JSX and why is it useful in React?
Answer: JSX (JavaScript XML) is a syntax extension for JavaScript that lets you write HTML-like structures directly inside JavaScript code.
Why it is useful:
	Readability: Makes code cleaner and visually resembles the resulting UI structure.
	Component-Centric: Keeps rendering logic and dynamic JavaScript logic together in one place.
	Type Safety & Security: Compiles to standard React.createElement() calls, automatically escaping strings to help prevent Cross-Site Scripting (XSS) attacks.


Question 4 : Explain how props support React’s one-way data flow.
Answer: In React, props (short for properties) flow exclusively in one direction: down from parent components to child components.
	Read-Only: Props are immutable; a child component cannot alter the props it receives.
	Predictability: Because children cannot directly change parent data, state management remains single-	sourced, predictable, and easier to debug.

Question 5 : Create a functional component ClickMeButton that renders a button. It accepts onClick and label props. When clicked, it executes the passed onClick handler.
Answer:  import React from 'react';

function ClickMeButton({ onClick, label }) {
  return (
    <button onClick={onClick}>
      {label}
    </button>
  );
}

export default ClickMeButton;
// Example usage:
function App() {
  const handleClick = () => {
    alert("Button was clicked!");
  };

  return (
    <div>
      <ClickMeButton onClick={handleClick} label="Click Me!" />
    </div>
  );
}
Question 6 : Given const fruits = ['Apple', 'Banana', 'Cherry'], write a component FruitList that renders an unordered list of these fruits using .map() and includes a key.
Answer: import React from 'react';

const fruits = ['Apple', 'Banana', 'Cherry'];

function FruitList() {
  return (
    <ul>
      {fruits.map((fruit, index) => (
        <li key={fruit}>{fruit}</li>
      ))}
    </ul>
  );
}

export default FruitList;


Question 7 : Why is the key attribute important when rendering lists in React? What goes wrong if keys are missing or misuse array index?
Answer: The key attribute gives React elements a stable identity across renders. React’s diffing algorithm uses keys to track which items in a list have been modified, added, or removed.
What Goes Wrong If Keys Are Missing or Array Indices Are Misused
	Missing Keys: React raises a browser console warning and falls back to using the array index as a key. This can lead to inefficient DOM updates because React cannot 	accurately track dynamic changes to list items.
	Using Array Indices as Keys: If the order of list items changes (e.g., sorting, filtering, inserting, or deleting items from the middle/top), using indices causes s	erious UI bugs:
		State Corruption: Child components with local state (like input fields or toggles) will retain state based on index position rather than item identity, 		assigning old state data to new elements.
		Performance Degradation: React fails to reorder elements efficiently, causing unnecessary DOM teardowns and re-creations instead of re-ordering existing 		nodes.
Best Practice: Always use a unique, stable ID from your data (such as a database ID) as the key.

Question 8 : Write a component UserList that receives a users prop—an array of objects each with id and name. Render each user’s name in a
tag, using map() and key.
Answer: import React from 'react';

function UserList({ users }) {
  return (
    <div>
      {users.map((user) => (
        <div key={user.id}>
          {user.name}
        </div>
      ))}
    </div>
  );
}

export default UserList;


Question 9 : Create an App component that holds an array of numbers, passes them to a NumberList child component for rendering, and passes a click handler function to log the clicked number. Show one-way data flow and event handling.
Answer: import React from 'react';

// Child Component
function NumberList({ numbers, onNumberClick }) {
  return (
    <ul>
      {numbers.map((num) => (
        <li key={num} onClick={() => onNumberClick(num)}>
          {num}
        </li>
      ))}
    </ul>
  );
}

// Parent Component
function App() {
  const numberArray = [10, 20, 30, 40, 50];

  const handleNumberClick = (selectedNumber) => {
    console.log(`Clicked number: ${selectedNumber}`);
  };

  return (
    <div>
      <h2>Number List</h2>
      {/* Passing state down via props (One-way data flow) and event callback */}
      <NumberList numbers={numberArray} onNumberClick={handleNumberClick} />
    </div>
  );
}

export default App;

Question 10 : Design a component ProductList that receives products (array of objects with id, name, price). Render a list of product names and prices using .map(). Each item when clicked should call a prop onProductSelect(product). Include the proper key.
Answer:import React from 'react';

function ProductList({ products, onProductSelect }) {
  return (
    <ul>
      {products.map((product) => (
        <li
          key={product.id}
          onClick={() => onProductSelect(product)}
          style={{ cursor: 'pointer' }}
        >
          {product.name} - ${product.price}
        </li>
      ))}
    </ul>
  );
}

export default ProductList;

'''
```
