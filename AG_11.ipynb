```python
'''
Question 1: Write a basic Node.js script that prints "Hello, World!" to the console and show how to execute it.
Answer: App.js
	console.log("Hello, World!");

	To Execute, Open your terminal or command prompt, navigate to the directory containing app.js,
	and run: node app.js


Question 2: Create a simple CommonJS module that exports a function to add two numbers. Then, create a separate script that imports and uses this function.
Answer: 1. Module File (math.js)
// Function to add two numbers
function add(a, b) {
  return a + b;
}

// Exporting the function using CommonJS syntax
module.exports = add;

	2. Main File (index.js)
	// Importing the exported function from math.js
const add = require("./math");

const result = add(5, 7);
console.log(`The sum is: ${result}`); // Output: The sum is: 12

Question 3: Show the commands to initialize a new Node.js project and install the axios package as a dependency.
Answer: # Initialize a new Node.js project with default settings (-y skips prompts)
	npm init -y

	# Install the axios package as a dependency
	npm install axios

Question 4: Write a script that accesses and displays the command-line arguments passed to it when executed.
Answer: JavaScript Code (args.js)
// process.argv contains command-line arguments
// process.argv[0] is the path to Node executable
// process.argv[1] is the path to the script being run
// process.argv.slice(2) extracts user-passed arguments

const userArgs = process.argv.slice(2);

console.log("All arguments received:", process.argv);
console.log("User arguments passed:", userArgs);
	Execution :
	node args.js hello 123 test

Question 5: Using the fs/promises module, write a script to asynchronously read the content of a file named data.txt and log it to the console.
Answer: const fs = require('fs/promises');

async function readDataFile() {
  try {
    // Asynchronously read the file using UTF-8 encoding
    const content = await fs.readFile('data.txt', 'utf-8');
    console.log("File content:\n", content);
  } catch (error) {
    console.error("Error reading file:", error.message);
  }
}

readDataFile();

Question 6: Using the fs/promises module, write a script to create a new file named log.txt and write the string "Task completed." into it.
Answer:
const fs = require('fs/promises');

async function createLogFile() {
  try {
    // Asynchronously writes text to log.txt (creates the file if it doesn't exist)
    await fs.writeFile('log.txt', 'Task completed.', 'utf-8');
    console.log('File log.txt was created successfully!');
  } catch (error) {
    console.error('Error writing to file:', error.message);
  }
}

createLogFile();

Question 7: Use the path module to safely join the segments src, utils, and helpers.js into a single, cross-platform-compatible file path.
Answer: const path = require('path');

// Safe, cross-platform path joining
const fullPath = path.join('src', 'utils', 'helpers.js');

console.log('Joined Path:', fullPath);
// On Windows: src\utils\helpers.js
// On Linux/macOS: src/utils/helpers.js

Question 8: Create a basic HTTP server using the http module that listens on port 3000 and responds to all requests with "Server is running."
Answer: const http = require('http');

const PORT = 3000;

// Create HTTP server
const server = http.createServer((req, res) => {
  res.writeHead(200, { 'Content-Type': 'text/plain' });
  res.end('Server is running.');
});

// Start listening on port 3000
server.listen(PORT, () => {
  console.log(`Server is listening on http://localhost:${PORT}`);
});


Question 9: Enhance the previous HTTP server to handle two different routes: / should respond with "Homepage", and /about should respond with "About Us".
Answer: const http = require('http');

const PORT = 3000;

const server = http.createServer((req, res) => {
  // Set response header
  res.writeHead(200, { 'Content-Type': 'text/plain' });

  // Handle simple routing based on URL path
  if (req.url === '/') {
    res.end('Homepage');
  } else if (req.url === '/about') {
    res.end('About Us');
  } else {
    res.writeHead(404, { 'Content-Type': 'text/plain' });
    res.end('404 Not Found');
  }
});

server.listen(PORT, () => {
  console.log(`Server running at http://localhost:${PORT}`);
});

Question 10: Write a script that efficiently reads a large file (large.txt) and outputs its content to the console using streams.
Answer: const fs = require('fs');

// Create a readable stream for memory-efficient chunk-by-chunk reading
const readStream = fs.createReadStream('large.txt', { encoding: 'utf-8' });

// Pipe the read stream directly to process.stdout (console)
readStream.pipe(process.stdout);

// Error handling
readStream.on('error', (error) => {
  console.error('Error reading the file stream:', error.message);
});

'''
```
