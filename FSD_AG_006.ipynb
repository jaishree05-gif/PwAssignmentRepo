```python
'''
uestion 1 : Explain what the forEach method does in JavaScript arrays and write code that uses forEach to display each student's name from an array of student objects to the console. // Array of student objects const students = [ { name: "Rahul", age: 20 }, { name: "Priya", age: 19 }, { name: "Amit", age: 21 }, { name: "Sneha", age: 20 } ];
Answer: The forEach() method in JavaScript is used to execute a function once for each element in an array. It does not create a new array or return a value. It is mainly used for displaying data, logging values, or performing actions on each element.
	// Array of student objects
	const students = [
  	{ name: "Rahul", age: 20 },
 	{ name: "Priya", age: 19 },
  	{ name: "Amit", age: 21 },
  	{ name: "Sneha", age: 20 }
	];

	// Display each student's name
	students.forEach(student => {
  		console.log(student.name);
	});
	output: [118, 295, 590, 885, 1180]

Question 2 : You have an array of product prices in rupees. Write code using the map method to create a new array that contains the same prices but with 18% GST added to each price. const originalPrices = [100, 250, 500, 750, 1000];
anser: const originalPrices = [100, 250, 500, 750, 1000];

	// Add 18% GST to each price
	const pricesWithGST = originalPrices.map(price => price * 1.18);

	console.log(pricesWithGST);

Question 3 : A school wants to find all students who scored more than 80 marks in their exam. The following code has an error that prevents it from working correctly. Fix the code and explain what was wrong. const students = [ { name: "Raj", marks: 85 }, { name: "Sita", marks: 92 }, { name: "Vikram", marks: 78 }, { name: "Meera", marks: 88 } ];
const topStudents = students.filter(student => { student.marks > 80; });
Answer:const students = [
  		{ name: "Raj", marks: 85 },
  		{ name: "Sita", marks: 92 },
  		{ name: "Vikram", marks: 78 },
  		{ name: "Meera", marks: 88 }
	];

	// Option 1: Explicit return statement
	const topStudents = students.filter(student => {
 		 return student.marks > 80;
	});

	// Option 2: Implicit return (shorter & cleaner)
	const topStudentsImplicit = students.filter(student => student.marks > 80);
	In the original code:
	Missing return statement: When you use curly braces {} in an arrow function (a block body), JavaScript expects an explicit return 	keyword.
	Returns undefined: Without return, the function evaluates student.marks > 80, discards the result, and returns undefined by default.
	Empty Result Array: Since undefined is falsy, filter() treats every student as a failed check, returning an empty array [] instead of 	filtering the actual top students.

Question 4 : Write code using the find method to locate a book with the title "Harry Potter" from an array of book objects. If the book is found, display its author name, otherwise display "Book not found". // Array of book objects const books = [ { title: "Gitanjali", author: "Rabindranath Tagore" }, { title: "Harry Potter", author: "J.K. Rowling" }, { title: "The God of Small Things", author: "Arundhati Roy" }, { title: "Midnight's Children", author: "Salman Rushdie" } ];
Answer: const books = [
  		{ title: "Gitanjali", author: "Rabindranath Tagore" },
  		{ title: "Harry Potter", author: "J.K. Rowling" },
  		{ title: "The God of Small Things", author: "Arundhati Roy" },
  		{ title: "Midnight's Children", author: "Salman Rushdie" }
	];

	// Locate the book
	const bookFound = books.find(book => book.title === "Harry Potter");

	// Check if found and display result
	if (bookFound) {
  		console.log(`Author: ${bookFound.author}`);
	} else {
  		console.log("Book not found");
	}

Question 5 : Explain how the sort method works in JavaScript and write code to sort an array of employee objects by their salary in descending order (highest salary first). // Array of employee objects const employees = [ { name: "Rajesh", salary: 45000 }, { name: "Priya", salary: 55000 }, { name: "Amit", salary: 40000 }, { name: "Kavya", salary: 60000 } ];
Answer: In-place Mutation: The .sort() method modifies (mutates) the original array directly rather than creating a new one.
	Default Behavior: By default, .sort() converts array elements into strings and compares them lexicographically (alphabetically). This 	works fine for strings, but causes bugs with numbers (e.g., "100" comes before "2").
	Compare Function (a, b): To sort numbers or objects reliably, you pass a comparator function with two parameters, a and b:

	Return < 0: Keeps a before b.
	Return > 0: Puts b before a.
	Return 0: Keeps original order relative to each other.

	To sort numbers in descending order (highest first), you evaluate b - a.
	const employees = [
  		{ name: "Rajesh", salary: 45000 },
  		{ name: "Priya", salary: 55000 },
  		{ name: "Amit", salary: 40000 },
  		{ name: "Kavya", salary: 60000 }
	];

	// Sort employees by salary in descending order
	employees.sort((a, b) => b.salary - a.salary);

	console.log(employees);
   	Output :[
  			{ name: "Kavya", salary: 60000 },
  			{ name: "Priya", salary: 55000 },
  			{ name: "Rajesh", salary: 45000 },
  			{ name: "Amit", salary: 40000 }
		]

Question 6 : Write code using the reduce method to calculate the total value of all items
in a shopping cart. Each item has a price and quantity. Also find the most expensive
single item in the cart.
// Shopping cart with items
const cart = [
{ name: "Laptop", price: 50000, quantity: 1 },
{ name: "Mouse", price: 500, quantity: 2 },
{ name: "Keyboard", price: 1500, quantity: 1 },
{ name: "Monitor", price: 15000, quantity: 1 }
];
Answer:
const cart = [
  { name: "Laptop", price: 50000, quantity: 1 },
  { name: "Mouse", price: 500, quantity: 2 },
  { name: "Keyboard", price: 1500, quantity: 1 },
  { name: "Monitor", price: 15000, quantity: 1 }
];

// 1. Calculate the total value of all items in the cart
const totalValue = cart.reduce((total, item) => {
  return total + item.price * item.quantity;
}, 0);

// 2. Find the most expensive single item (by unit price)
const mostExpensiveItem = cart.reduce((maxItem, currentItem) => {
  return currentItem.price > maxItem.price ? currentItem : maxItem;
}, cart[0]);

console.log(`Total Cart Value: ₹${totalValue}`);
console.log(`Most Expensive Item: ${mostExpensiveItem.name} (Price: ₹${mostExpensiveItem.price})`);

Question 7 : Explain what array destructuring is and write code that extracts the first
two colors from an array of colors into separate variables using destructuring.
// Array of colors
const colors = ["red", "blue", "green", "yellow", "purple"];
Answer:
Array destructuring is a feature in ES6 JavaScript that allows you to unpack values from an array into distinct variables using a clean, concise syntax that mirrors the structure of an array literal.

Instead of accessing elements manually using index numbers (e.g., const first = colors[0]), destructuring lets you extract multiple elements in a single line.
// Array of colors
const colors = ["red", "blue", "green", "yellow", "purple"];

// Extracting the first two colors into separate variables
const [firstColor, secondColor] = colors;

console.log(firstColor);  // "red"
console.log(secondColor); // "blue"

Question 8 : You have a student object with name, age, and marks properties. Write
code using object destructuring to extract these values into separate variables and also
rename the marks property to score while destructuring.
// Student object
const student = {
name: "Arjun",
age: 20,
marks: 85,
city: "Mumbai"
};
Answer:
// Student object
const student = {
  name: "Arjun",
  age: 20,
  marks: 85,
  city: "Mumbai"
};

// Object destructuring: extracting values and renaming 'marks' to 'score'
const { name, age, marks: score } = student;

console.log(name);  // "Arjun"
console.log(age);   // 20
console.log(score); // 85

Question 9 : Write a function called calculateTotal that can accept any number of
prices as arguments and returns their sum. Use the rest operator to handle the variable
number of parameters.
Answer:

function calculateTotal(...prices) {
  return prices.reduce((sum, price) => sum + price, 0);
}

// Example usage:
console.log(calculateTotal(10, 20, 30));       // 60
console.log(calculateTotal(100, 250, 50, 75)); // 475
console.log(calculateTotal());                 // 0

Question 10 : You have two arrays of student names and need to create a new array
that contains all students from both arrays. Also, you have two objects representing a
student's basic info and marks. Write code using the spread operator to solve both
scenarios.
const classA = ["Rahul", "Priya", "Amit"];
const classB = ["Sneha", "Vikram", "Pooja"];
Answer:
// Data given
const classA = ["Rahul", "Priya", "Amit"];
const classB = ["Sneha", "Vikram", "Pooja"];

const studentInfo = { name: "Rohan", age: 21, city: "Delhi" };
const studentMarks = { math: 90, science: 85, english: 88 };

// Scenario 1: Merge two arrays of student names
const allStudents = [...classA, ...classB];

console.log(allStudents);
// Output: ["Rahul", "Priya", "Amit", "Sneha", "Vikram", "Pooja"]


// Scenario 2: Merge basic info and marks objects into a single student object
const fullStudentDetails = { ...studentInfo, ...studentMarks };

console.log(fullStudentDetails);
// Output: { name: "Rohan", age: 21, city: "Delhi", math: 90, science: 85, english: 88 }



'''
```
