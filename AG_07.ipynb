```python
'''
Question 1 : Write code that creates a new paragraph element with the text "Welcome to our website!" and adds it to a div with the id "content" on the webpage. Explain what createElement and appendChild methods do.
Answer:
 // Create a new paragraph element
const newParagraph = document.createElement("p");

// Add the text content
newParagraph.textContent = "Welcome to our website!";

// Find the div with the id "content" and append the paragraph
const contentDiv = document.getElementById("content");
contentDiv.appendChild(newParagraph);

document.createElement(tagName): Creates a new HTML element node specified by the tag name (e.g., "p", "div"). At this point, the element exists only in memory and is not yet visible on the webpage.
appendChild(childNode): Takes an existing element node and attaches it as the last child of a specified parent element. This step actually inserts the element into the DOM, making it appear on the webpage.

Question 2 : You need to change the appearance of a button when users hover over it. Write code that finds a button with id "myButton" and changes its background color to blue and text color to white using the element.style property.
Answer: const button = document.getElementById("myButton");

// Change styles when the mouse enters (hover)
button.addEventListener("mouseover", () => {
  button.style.backgroundColor = "blue";
  button.style.color = "white";
});

// Optional: Reset styles when the mouse leaves
button.addEventListener("mouseout", () => {
  button.style.backgroundColor = "";
  button.style.color = "";
});

Question 3 : Create a button that shows an alert with the message "Button was clicked!" when clicked. Also, add a second event listener to the same button that changes its text to "Clicked!" after being clicked. Explain why addEventListener is better than using onclick.
Answer: const button = document.getElementById("myButton");

// First event listener: Show an alert
button.addEventListener("click", () => {
  alert("Button was clicked!");
});

// Second event listener: Change the button's text
button.addEventListener("click", () => {
  button.textContent = "Clicked!";
});

Multiple Handlers: addEventListener allows you to attach multiple event listeners to the exact same event on a single element. In contrast, using element.onclick = ... overwrites any previously assigned click handler.
Separation of Concerns & Flexibility: addEventListener works cleanly inside external JavaScript files without risking accidental overwrites from other scripts. It also supports advanced options like event propagation control (capturing vs. bubbling) and can easily be detached using removeEventListener.

Question 4 : Write a function named extractDomain that takes an email string like "john@example.com" and returns the domain part ("example.com") using only string methods. Do not use regular expressions.
Answer: function extractDomain(email) {
  // Find the index of the '@' character and extract everything after it
  const atIndex = email.indexOf("@");
  return email.slice(atIndex + 1);
}

// Example usage:
console.log(extractDomain("john@example.com")); // "example.com"


Question 5 : Write a function called repeatMessage that takes a message string and a number, and returns the message repeated that many times, separated by a space. Do not use loops. Example: repeatMessage("hello", 3) should return "hello hello hello"
Answer: function repeatMessage(message, count) {
  // Option 1: Using String.prototype.repeat() and trim()
  return (message + " ").repeat(count).trim();

  // Option 2 (Alternative): Using Array.fill() and join()
  // return Array(count).fill(message).join(" ");
}

// Example usage:
console.log(repeatMessage("hello", 3)); // "hello hello hello"


Question 6 : Explain what the Event Loop is in JavaScript and predict the output of the following code. Also explain why the output appears in that specific order. console.log('Start'); setTimeout(function() { console.log('Timeout 1'); }, 0); console.log('Middle'); setTimeout(function() { console.log('Timeout 2'); }, 0); console.log('End');
Answer: JavaScript is a single-threaded programming language, meaning it can only execute one task at a time on its Call Stack. The Event Loop is the underlying mechanism that handles asynchronous operations (like setTimeout, fetch calls, or event listeners).
It continuously monitors two main areas:
	Call Stack: Where current execution takes place.
	Callback Queue (Task Queue): Where asynchronous callbacks wait after their triggers/timers complete.
If the Call Stack is empty, the Event Loop takes the first callback waiting in the Callback Queue and pushes it onto the Call Stack to execute.

2. Predicted Output
Start
Middle
End
Timeout 1
Timeout 2

3. Why it Appears in That Order
	console.log('Start') runs synchronously and logs "Start".
	setTimeout(..., 0) is handed off to the browser's Web API. Even with a delay of 0 milliseconds, its callback is moved to the Callback Queue, not executed immediately.
	console.log('Middle') runs synchronously and logs "Middle".
	The second setTimeout(..., 0) hands off its callback, placing it second in line in the Callback Queue.
	console.log('End') runs synchronously and logs "End".
	Now that the Call Stack is completely clear, the Event Loop pulls the callbacks from the Callback Queue in FIFO (First-In, First-Out) order, logging "Timeout 1" 	followed by "Timeout 2".

Question 7 : Write a JavaScript function named startCountdown that logs “Time left: X” every second, starting from 5 down to 1, and finally logs “Time’s up!”. Use setInterval and clearInterval only.
Answer: function startCountdown() {
  let count = 5;

  const timerId = setInterval(() => {
    if (count > 0) {
      console.log(`Time left: ${count}`);
      count--;
    } else {
      console.log("Time’s up!");
      clearInterval(timerId); // Stop the interval
    }
  }, 1000);
}

// Call the function to test
startCountdown();

Question 8 : Explain what Promises are in JavaScript and write code that creates a Promise to simulate checking if a user is logged in. The Promise should resolve with user data after 2 seconds if successful, or reject with an error message if the user is not found.
Answer:  A Promise in JavaScript is an object representing the eventual completion (or failure) of an asynchronous operation and its resulting value. It acts as a placeholder for a value that is not necessarily known when the promise is created.
	A Promise exists in one of three states:
		Pending: Initial state, neither fulfilled nor rejected.
		Fulfilled (Resolved): The operation completed successfully.
		Rejected: The operation failed.
function checkUserLogin(isUserFound) {
  return new Promise((resolve, reject) => {
    setTimeout(() => {
      if (isUserFound) {
        resolve({
          id: 101,
          name: "Alex Smith",
          email: "alex@example.com"
        });
      } else {
        reject("Error: User not found.");
      }
    }, 2000);
  });
}

// Example usage:
checkUserLogin(true)
  .then(userData => console.log("Success:", userData))
  .catch(error => console.error(error));

Question 9 : Rewrite the following Promise-based code to use async/await syntax. The code fetches user data and then fetches that user's posts. Include proper error handling. function getUserDataAndPosts(userId) { return fetchUserData(userId) .then(userData => { return fetchUserPosts(userData.id) .then(posts => { return { user: userData, posts: posts }; }); }) .catch(error => { console.error('Error:', error); throw error; }); }
Answer: async function getUserDataAndPosts(userId) {
  try {
    const userData = await fetchUserData(userId);
    const posts = await fetchUserPosts(userData.id);

    return {
      user: userData,
      posts: posts
    };
  } catch (error) {
    console.error('Error:', error);
    throw error;
  }
}


Question 10 : Write code using the fetch API to get weather data from a public API. Display the weather information in a user-friendly format. Assume you have the following configuration for the API. const apiKey = 'abcd1234'; const apiUrl = `https://api.openweathermap.org/data/2.5/weather?q=Mumbai&appid=${apiK ey}&units=metric`;
Answer:  const apiKey = 'abcd1234';
const apiUrl = `https://api.openweathermap.org/data/2.5/weather?q=Mumbai&appid=${apiKey}&units=metric`;

async function getWeatherData() {
  try {
    const response = await fetch(apiUrl);

    // Check if the HTTP response status is OK (200-299)
    if (!response.ok) {
      throw new Error(`HTTP error! Status: ${response.status}`);
    }

    const data = await response.json();

    // Display formatted output
    console.log("=== Weather Report ===");
    console.log(`City: ${data.name}, ${data.sys.country}`);
    console.log(`Condition: ${data.weather[0].description}`);
    console.log(`Temperature: ${data.main.temp}°C (Feels like: ${data.main.feels_like}°C)`);
    console.log(`Humidity: ${data.main.humidity}%`);
    console.log(`Wind Speed: ${data.wind.speed} m/s`);
    console.log("======================");

  } catch (error) {
    console.error("Failed to fetch weather data:", error.message);
  }
}

// Execute the request
getWeatherData();



'''
```
