```python
'''
Question 1 : Create an object named student with the following properties: ● name: "Rahul" ● age: 21 ● courses: an array containing "Math", "Physics", and "Chemistry" Then, perform the following operations: 1. Add a new property grade with the value "A". 2. Update the age to 22. 3. Delete the courses property.
Answer:
// Initial Object
const student = {
  name: "Rahul",
  age: 21,
  courses: ["Math", "Physics", "Chemistry"]
};

// 1. Add a new property grade with value "A"
student.grade = "A";

// 2. Update the age to 22
student.age = 22;

// 3. Delete the courses property
delete student.courses;

console.log(student);
// Output: { name: 'Rahul', age: 22, grade: 'A' }

Question 2: Write a function checkEligibility that takes an age as a parameter and returns: ● "Eligible" if age is 18 or above ● "Not Eligible" if age is below 18 Use an if-else statement.
Answer:
function checkEligibility(age) {
  if (age >= 18) {
    return "Eligible";
  } else {
    return "Not Eligible";
  }
}

console.log(checkEligibility(20)); // Output: "Eligible"
console.log(checkEligibility(15)); // Output: "Not Eligible"

Question 3: Write a function getDayName that takes a number (1-7) and returns the corresponding day of the week. Use a switch statement.
Answer:
function getDayName(dayNumber) {
  switch (dayNumber) {
    case 1:
      return "Monday";
    case 2:
      return "Tuesday";
    case 3:
      return "Wednesday";
    case 4:
      return "Thursday";
    case 5:
      return "Friday";
    case 6:
      return "Saturday";
    case 7:
      return "Sunday";
    default:
      return "Invalid day number";
  }
}

console.log(getDayName(3)); // Output: "Wednesday"

Question 4: Rewrite the checkEligibility function from Question 2 using the ternary operator.
Answer:
const checkEligibility = (age) => (age >= 18 ? "Eligible" : "Not Eligible");

console.log(checkEligibility(18)); // Output: "Eligible"

Question 5: Given an array numbers = [1, 2, 3, 4, 5], perform the following: 1. Use a for loop to print each number. 2. Use a while loop to print each number. 3. Use a for-in loop to print each index. 4. Use a for-of loop to print each number.
Answer:
const numbers = [1, 2, 3, 4, 5];

// 1. standard 'for' loop
for (let i = 0; i < numbers.length; i++) {
  console.log(numbers[i]);
}

// 2. 'while' loop
let i = 0;
while (i < numbers.length) {
  console.log(numbers[i]);
  i++;
}

// 3. 'for-in' loop (iterates over array indices/keys)
for (const index in numbers) {
  console.log(index);
}

// 4. 'for-of' loop (iterates over array values)
for (const num of numbers) {
  console.log(num);
}

Question 6: Explain what functions are in JavaScript and provide an example of a function that adds two numbers.
A function in JavaScript is a reusable block of code designed to perform a specific task. Functions allow you to write code once and execute it multiple times with different inputs (arguments), helping to keep code modular and organized.
Answer:
function add(a, b) {
  return a + b;
}

console.log(add(5, 10)); // Output: 15

Question 7: Convert the add function from Question 6 into an arrow function.
Answer:
const add = (a, b) => a + b;

console.log(add(5, 10)); // Output: 15

Question 8: Write a function greet that takes a name and a callback function. The greet function should call the callback with the name.
Answer:
function greet(name, callback) {
  callback(name);
}

// Callback execution test
function printGreeting(personName) {
  console.log(`Hello, ${personName}!`);
}

greet("Rahul", printGreeting); // Output: Hello, Rahul!

Question 9: Given the array: const fruits = ["apple", "banana", "cherry"]; 1. What is the length of the array? 2. Add "date" to the end of the array using the length property.
Answer:
const fruits = ["apple", "banana", "cherry"];

// 1. Length of the array
console.log(fruits.length); // Output: 3

// 2. Add "date" to the end using the .length property
fruits[fruits.length] = "date";

console.log(fruits); // Output: ["apple", "banana", "cherry", "date"]

Question 10: Given the string text = " Hello World! ", perform the following: 1. Trim the whitespace. 2. Convert to uppercase. 3. Check if it includes “World".
const text = " Hello World! ";

// 1. Trim whitespace
const trimmedText = text.trim();
console.log(trimmedText); // Output: "Hello World!"

// 2. Convert to uppercase
const upperText = trimmedText.toUpperCase();
console.log(upperText); // Output: "HELLO WORLD!"

// 3. Check if it includes "World"
const hasWorld = trimmedText.includes("World");
console.log(hasWorld); // Output: true

Question 11: Given the array numbers = [1, 2, 3, 4, 5], perform the following: 1. Add 6 to the end. 2. Remove the first element. 3. Find the index of 4. 4. Check if all numbers are greater than 0.
Answer:
let numbers = [1, 2, 3, 4, 5];

// 1. Add 6 to the end
numbers.push(6); // numbers becomes [1, 2, 3, 4, 5, 6]

// 2. Remove the first element
numbers.shift(); // numbers becomes [2, 3, 4, 5, 6]

// 3. Find the index of 4
const indexOfFour = numbers.indexOf(4);
console.log(indexOfFour); // Output: 2

// 4. Check if all numbers are greater than 0
const allGreaterThanZero = numbers.every(num => num > 0);
console.log(allGreaterThanZero); // Output: true


'''
```
